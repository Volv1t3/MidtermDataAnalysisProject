### Hipótesis a comprobar

#### **Hipótesis 1: Comparación entre Plataformas ($\text{Xbox One}$ vs. $\text{PC}$)**

*   **$\text{H}_0$ (Hipótesis Nula):** **No hay diferencia estadísticamente significativa** en la calificación promedio de los usuarios entre juegos lanzados en Xbox One y PC.
    $$\text{H}_0: \mu_{\text{XboxOne}} = \mu_{PC}$$
*   **$\text{H}_1$ (Hipótesis Alternativa):** **Existe una diferencia estadísticamente significativa** en la calificación promedio de los usuarios entre juegos lanzados en Xbox One y PC.
    $$\text{H}_1: \mu_{\text{XboxOne}} \neq \mu_{PC}$$


#### **Hipótesis 2: Comparación entre Géneros ($\text{Acción}$ vs. $\text{Deportes}$)**

*   **$\text{H}_0$ (Hipótesis Nula):** **No hay diferencia estadísticamente significativa** en la calificación promedio de los usuarios entre juegos del género Acción y los del género Deportes.
    $$\text{H}_0: \mu_{\text{Acción}} = \mu_{\text{Deportes}}$$
*   **$\text{H}_1$ (Hipótesis Alternativa):** **Existe una diferencia estadísticamente significativa** en la calificación promedio de los usuarios entre juegos del género Acción y los del género Deportes.
    $$\text{H}_1: \mu_{\text{Acción}} \neq \mu_{\text{Deportes}}$$

```python
from typing import Any, Iterable
from seaborn.palettes import _ColorPalette


def crear_barchart_storytelling(
        data: pd.DataFrame,
        x_axis_data: Any | str | pd.Series,
        x_axis_title: str,
        y_axis_data: Any | str | pd.Series,
        y_axis_title: str,
        fig_size: tuple[int, int],
        color_palette: Any,
        title: str,
        subtitle: str,
        #? Orientacion
        orientation: str = 'horizontal',
        #? Grid
        grid_axis: str = 'x',
        grid_color: str = 'lightgray',
        grid_linestyle: str = '--',
        grid_alpha: float = 0.4,
        #? Spines
        visible_spines: set[str] = None,
        spine_colors: dict[str, str] = None,
        #? Bar labels
        show_bar_labels: bool = True,
        bar_label_offset: float = 10,
        bar_label_fontsize: int = 10,
        bar_label_color: str = 'gray',
        bar_label_formatter: callable = None,
        #? Hue y orden
        hue: str = None,
        hue_order: list[str] = None,
        order: list[str] = None,
        #? Ejes
        show_x_axis: bool = True,
        show_y_axis: bool = True,
        x_tick_rotation: float = 0,
        y_tick_rotation: float = 0,
        #? Leyenda
        show_legend: bool = False,
        legend_position: str = 'best',
        legend_title: str = None,
        show_all_y_axis_ticks: bool = True,
        show_all_x_axis_ticks: bool = True
) -> tuple[plt.Figure, plt.Axes]:
    """
    Funcion diseñada para generar una visualizacion orientada al storytelling con un titulo centrado, subtitulo con insight, y
    data programable para evitar la reescritura del codigo de las visualizaciones.

    Parameters:
        data (pd.DataFrame): The data for the bar chart.
        x_axis_data (Any): nombre de la columna o serie de datos correspondiente al eje x
        x_axis_title (str): titulo del eje x
        y_axis_data (Any): nombre de la columna o serie de datos correspondiente al eje y
        y_axis_title (str): titulo del eje y
        fig_size (tuple): tamano de la figura
        color_palette (Any): Paleta de colores para las barras
        title (str): Titulo del grafico
        subtitle (str): Subtitulo con el insight
        orientation (str): 'horizontal' o 'vertical' (default: 'horizontal')
        grid_axis (str): 'x', 'y', 'both', o 'none' (default: 'x')
        grid_color (str): color de la grid (default: 'lightgray')
        grid_linestyle (str): estilo de linea de la grid (default: '--')
        grid_alpha (float): transparencia de la grid (default: 0.4)
        visible_spines (set): set de spines visibles e.g. {'bottom', 'left'} (default: {'bottom'})
        spine_colors (dict): colores por spine e.g. {'bottom': 'lightgray'} (default: {'bottom': 'lightgray'})
        show_bar_labels (bool): mostrar etiquetas sobre las barras (default: True)
        bar_label_offset (float): distancia de la etiqueta respecto a la barra (default: 5)
        bar_label_fontsize (int): tamano de fuente de las etiquetas (default: 10)
        bar_label_color (str): color de las etiquetas (default: 'gray')
        bar_label_formatter (callable): funcion para formatear el valor e.g. lambda v: f'{v:.2f}%' (default: str)
        hue (str): columna para agrupar barras por color (default: None)
        hue_order (list): orden explicito de las categorias del hue (default: None)
        order (list): orden explicito de las categorias del eje categorico (default: None)
        show_x_axis (bool): mostrar eje x (default: True)
        show_y_axis (bool): mostrar eje y (default: True)
        x_tick_rotation (float): rotacion de las etiquetas del eje x (default: 0)
        y_tick_rotation (float): rotacion de las etiquetas del eje y (default: 0)
        show_legend (bool): mostrar leyenda (default: False)
        legend_position (str): posicion de la leyenda (default: 'best')
        legend_title (str): titulo de la leyenda (default: None)
        show_all_y_axis_ticks (bool): si se muestran todas las ticks del eje y (default True),
        show_all_x_axis_ticks (bool): si se muestran todas las ticks del eje x (default True)


    Returns:
        tuple[plt.Figure, plt.Axes]: figura y ejes del grafico generado
    """

    #? 0. Defaults para parametros mutables
    #? Aqui cargamos las variables de las splines, o las lineas de la cuadricula que se mostraran dependiendo de los ejes que se solicitan en la funcion
    if visible_spines is None:
        if show_y_axis and show_x_axis:
            visible_spines = {'left', 'bottom'}
        elif show_x_axis:
            visible_spines = {'bottom'}
        else:
            visible_spines = {'left'}
    #? Aqui le anadimos los colores de las spines dependiendo de lo que queramos resaltar en base de los ejes que se muestran
    if spine_colors is None:
        if show_y_axis and show_x_axis:
            spine_colors = {'bottom': 'lightgray', 'left':'lightgray'}
        elif show_x_axis:
            spine_colors = {'bottom': 'lightgray'}
        else:
            spine_colors = {'left': 'lightgray'}
    #? Si no tenemos un label formatter, nuestra funcion lambda para el formato corresponde al mismo valor
    if bar_label_formatter is None:
        bar_label_formatter = lambda v: f'{v}'


    #? 1. Definimos el tipo de grafico como barplot en su propio subplot 1x1
    fig, ax = plt.subplots(figsize=fig_size)

        #? 1.1 Dibujamos las barras con matplotlib directamente y ordenamos la data si es que el eje es categorico
    plot_data = data.copy()
    plot_colors = color_palette

    #? 1.1.a Determinamos cual eje es categorico segun la orientacion
    #? Si es un bar chart entonces esperamos que las categorias esten en la y y los datos en la x
    if orientation == 'horizontal':
        cat_axis = y_axis_data
        val_axis = x_axis_data
    #? Si es un colum chart ersperamos alreves
    else:
        cat_axis = x_axis_data
        val_axis = y_axis_data

    #? Guardamos las categortias y los valores en sus variables respectivas para usar
    #? en los graficos
    categories = plot_data[cat_axis] if isinstance(cat_axis, str) else cat_axis
    values = plot_data[val_axis] if isinstance(val_axis, str) else val_axis


    #? Creamos el grafico especifico dependiendo de la orientacion
    if orientation == 'horizontal':
        bars = ax.barh(categories, values, color=hue_order)
    else:
        bars = ax.bar(categories, values, color=hue_order)



    #? 2. Le anadimos las labels requeridas al grafico
    ax.set_title(title, fontsize=16, fontweight='bold', loc='left', color='black', pad=45)
    ax.text(0.0, 1.01, subtitle, transform=ax.transAxes,
            fontsize=10, fontstyle='italic', color='gray', va='bottom', ha='left')
    #? 2.1 Le anadimos las labels a los ejes
    if x_axis_title is not None and isinstance(x_axis_title, str):
        ax.set_xlabel(x_axis_title, fontdict={'fontsize': 9, 'fontweight': 'normal'}, labelpad=15, ha='left', x=0)
    if y_axis_title is not None and isinstance(y_axis_title, str):
        ax.set_ylabel(y_axis_title, fontdict={'fontsize': 9, 'fontweight': 'normal'}, labelpad=15, ha='center', va='bottom', y=0.2)

    #? 3. Configuracion de la grid
    if grid_axis == 'none':
        ax.grid(False)
    elif grid_axis == 'both':
        ax.grid(True, color=grid_color, linestyle=grid_linestyle, alpha=grid_alpha)
    else:
        ax.grid(False)
        ax.grid(True, axis=grid_axis, color=grid_color, linestyle=grid_linestyle, alpha=grid_alpha)

    #? 4. Configuracion de spines
    for spine in ['top', 'right', 'left', 'bottom']:
        ax.spines[spine].set_visible(spine in visible_spines)
        if spine in spine_colors:
            ax.spines[spine].set_color(spine_colors[spine])

    #? 5. Visibilidad de ejes
    ax.xaxis.set_visible(show_x_axis)
    ax.yaxis.set_visible(show_y_axis)

    #? 5.1 Rotacion de ticks
    if x_tick_rotation != 0:
        ax.tick_params(axis='x', rotation=x_tick_rotation)
    if y_tick_rotation != 0:
        ax.tick_params(axis='y', rotation=y_tick_rotation)

     #? 5.2 Ticks de los ejes
    if show_all_y_axis_ticks:
        if isinstance(y_axis_data, str):
            y_axis_ticks = data[y_axis_data].tolist()
        else:
            y_axis_ticks = y_axis_data.tolist()
        ax.set_yticks(y_axis_ticks)
        ax.set_yticklabels([str(v) for v in y_axis_ticks], fontsize=10)
    if show_all_x_axis_ticks:
        if isinstance(x_axis_data, str):
            x_axis_ticks = data[x_axis_data].tolist()
        else:
            x_axis_ticks = x_axis_data.tolist()
        ax.set_xticks(x_axis_ticks)
        ax.set_xticklabels([str(v) for v in x_axis_ticks], fontsize=10)



    #? 6. Etiquetas sobre las barras
    if show_bar_labels:
        if orientation == 'horizontal':
            ax.set_xlim(right=ax.get_xlim()[1] * 1.15)
        else:
            ax.set_ylim(top=ax.get_ylim()[1] * 1.15)

        for bar in bars:
            if orientation == 'horizontal':
                width = bar.get_width()
                if width != 0:
                    ax.text(width, bar.get_y() + bar.get_height() / 2,
                            bar_label_formatter(width), color=bar_label_color,
                            va='center', fontsize=bar_label_fontsize)
            else:
                height = bar.get_height()
                if height != 0:
                    ax.text(bar.get_x() + bar.get_width() / 2, height,
                            bar_label_formatter(height), color=bar_label_color,
                            ha='center', fontsize=bar_label_fontsize, va='bottom')



    #? 7. Leyenda
    if show_legend and hue is not None:
        ax.legend(title=legend_title, loc=legend_position)
    elif not show_legend:
        legend = ax.get_legend()
        if legend is not None:
            legend.remove()

    plt.tight_layout(pad=2.0)
    plt.show()

def generate_blue_palette(n: int) -> _ColorPalette:
    """
    Funcion encargada de generar una paleta de colores replicable en base de la paleta built-in de Blues de Seaborn
    y usando el parametro n para determinar cuantos colores generar y retornar para la visualizacion
    :param n: entero representativo de la cantidad de colores necesarios
    :return: listado de colores
    """
    return sns.color_palette("Blues", n)

clasificacion_binaria_extraida = pd.DataFrame(data_to_explore['Clasificacion de Nulos con respecto a Year of Release'].value_counts())
crear_barchart_storytelling(
    data=clasificacion_binaria_extraida.reset_index(),
    y_axis_data="Clasificacion de Nulos con respecto a Year of Release",
    x_axis_title='Clasificacion Binaria',
    x_axis_data="count",
    y_axis_title='Coneo de Nulos',
    fig_size=(10,6),
    color_palette=generate_blue_palette(len(clasificacion_binaria_extraida.index)),
    title="Conteo de Nulos por Columna Nula y Combinaciones",
    subtitle="El conteo demuestra que la combinacion mas alta corresponde a 6k columnas nulas en donde Rating y ambos scores son nulos,\nlo que indica que tenemos un numero importante de juegos sin calificaciones de usuarios ni criticos, lo que podria indicar un sesgo\nen los datos hacia juegos menos populares o con menor impacto comercial.",
    show_all_x_axis_ticks=False, show_all_y_axis_ticks=True)
```

Como se puede observar en base del resultado de la revisión de nulos del dataset, el problema más grande de los datos es que tenemos un core de solo 7878 datos correctos (se asume que las columnas de ventas siempre están correctas dado que un análisis de sus datos demuestra que tienen todos los valores registrados), lo que corresponde a un 47% de los datos originales que podemos considerar nuestro *core analítico fuerte*, A la par de esto notamos que existen 6534 entradas en donde los nulos son las variables clasificatorias de los juegos, es decir, user rating, critic score y rating, lo que corresponde al 39% de los datos reales que en este caso tendrían un problema de registro de datos.

```python
import re

def normalize_token(text: str) -> str:
    """
    Funcion usada para la correccion de caracteres extranos en cadenas de texto en el dataset, lo usamos para eliminar
    espacios, apostrofes, dashes, emdashes, etc.
    :param text:
    :return:
    """

    if text is None:
        return text

    s = str(text).lower().strip()
    s = s.replace('/', '_')
    s = s.replace(' ','_')
    s = s.strip('_')
    return s
```

```python
#? 4. Correccion de la columna de Name, Platform y Genre, con una normalizacion de nombres basados en una eliminacion de caracteres que no sean numeros o letras, y su reemplazo con espacios en una primera pasada para luego reemplazarlos con dashes

def normalize_name_improved(text: str)-> str:
    if text is None:
        return text

    #? Realizamos una limpieza de cada palabra dentro de la lista obtenida al romper por espacios primero
    parts_lower_cased = [str.lower(part) for part in text.split(' ')]

    for idx, part_lowered in enumerate(parts_lower_cased):
        parts_lower_cased[idx] = re.sub(r'[^a-z0-9\s]', '', part_lowered.lower().strip())

    #? Retornamos con nuevo separador
    return '_'.join(parts_lower_cased)


critical_text_columns = ['Name','Platform','Genre']

for column in critical_text_columns:
    data_cleaned[column] = data_cleaned[column].apply(func=normalize_name_improved)

data_cleaned
```

```python
#? 1. Separamos registros validos, es decir aquellos que no son -1 y aquellos que son erroneos que son -1


registros_validos = data_cleaned[data_cleaned['User_Score'] > 0].copy()
registros_invalidos = data_cleaned[data_cleaned['User_Score'] < 0].copy()


#? 1.1 Exploracion inicial por conteo de los valores registraods
print(f"Registros validos (positivos): {len(registros_validos)}")
print(f"Registros erroneos (negativos): {len(registros_invalidos)}")

#? 1.2 Calculamos la mediana por grupo usando solo datos validos
columnas_pivote = ['Platform', 'Year_of_Release', 'Genre']

#? 1.2.1 En este caso, calculamos la media y el conteo dado que en primera instancia necesitamos analizar todos los posibles groups, en este caso 201, y sus valores de mediana. Esta es la base para el left join con la tabla de valores incorrectos
medianas_referenciales_para_determinancia_correccion = (registros_validos
                                                        .groupby(columnas_pivote)['User_Score']
                                                        .agg(['median', 'count'])
                                                        .rename(columns={'median': 'Mediana Score', 'count': 'Conteo Referencia'}))

#? 1.3. Verificamos cobertura sobre positivos para determinar quien tiene un grupo en la tabla de referencia.
#? Para esto hacemos un merge, como Power Bi Merge Query con un tipo left join con las
#? columnas pivote, o de la llave compuesta de gender, family_status y education. En este caso, los parametros usados determinan lo siguiente
#* a) registros_invalidos.merge() realiza una operacione stilo power query merge con base en los registros invalidso junto a una tabla a la derecha,
#* b) medianas_referenciales_para_determinancia_correcion tiene un indice compuesto de tres columnas, las columnas pivote, y en este caso es la izquierda para el left join.
#* c) el how='left' determina que el estilo de merge es un merge de tipo sql left join donde se deja nulo las combinaciones que no tienen un grupo al que pueden ser unidos en el lado derecho.
#* d) right_index determina que se tiene que usar el indice como lookup para determinar relaciones entre la izqueirda y la derecha, en este caso, se usa las columans pivote para la parte izquierda y el indice compuesto en la derecha
invalid_with_ref = registros_invalidos.merge(
    medianas_referenciales_para_determinancia_correccion, left_on=columnas_pivote, right_index=True, how='left'
)


print(f"\nCorregibles: {invalid_with_ref['Mediana Score'].notna().sum()} de {len(registros_invalidos)}")
print(f"Sin grupo de referencia: {invalid_with_ref['Mediana Score'].isna().sum()}")

#? 1.4. Vista previa de los grupos de referencia
medianas_referenciales_para_determinancia_correccion.sort_values('Conteo Referencia', ascending=False)
```

```python
#? 4. Aplicamos la correccion reemplazando valores NaN y -1 (tbd) con la mediana de su grupo para USER SCORE

registros_validos_us = data_cleaned[data_cleaned['User_Score'] > 0].copy()

#? 4.1 Aqui hacemos el proceso de agrupacion de nuevo pero con la idea de usar solo los dias validos y obtener la mediana en base a la misma agrupacion anterior, solo que esta agrupacion se usa directamente en el mecanismo de correccion mediante lambdas
medianas_finales_us = registros_validos_us.groupby(by=columnas_pivote)['User_Score'].median()

#? 4.2 Corregimos los valores dependiendo de si el valor es NaN o negativo (-1 de tbd), si lo es entonces buscamos su valor dentro del registro de medianas por grupo, si no lo es entonces copiamos el valor original.
data_cleaned['User_Score'] = data_cleaned.apply(
    lambda row:
        medianas_finales_us.get(
            tuple(row[col] for col in columnas_pivote))
        if (pd.isna(row['User_Score']) or row['User_Score'] < 0) else row['User_Score'],
    axis=1
)

print(f"{'Resultados de Correccion para User_Score':=^120}")
print(f"Valores positivos restantes: {(data_cleaned['User_Score'] > 0).sum()}")
print(f"Valores negativos restantes: {(data_cleaned['User_Score'] < 0).sum()}")
print(f"Valores NaN restantes: {data_cleaned['User_Score'].isna().sum()}")
print(f"Total: {len(data_cleaned)}")
```

```python
#? 6. Separamos registros validos, es decir aquellos que tienen Rating y aquellos que no

registros_validos = data_cleaned[data_cleaned['Rating'].notna()].copy()
registros_invalidos = data_cleaned[data_cleaned['Rating'].isna()].copy()

#? 6.1 Exploracion inicial por conteo de los valores registrados
print(f"Registros validos: {len(registros_validos)}")
print(f"Registros sin Rating: {len(registros_invalidos)}")

#? 6.2 Calculamos la moda por grupo usando solo datos validos
columnas_pivote = ['Platform', 'Year_of_Release', 'Genre']

#? 6.2.1 En este caso, calculamos la moda y el conteo. La moda se obtiene con una lambda
#? porque .mode() retorna una Serie (puede haber empates), asi que tomamos el primer valor con .iloc[0]
modas_referenciales_para_determinancia_correccion = (registros_validos
                                                     .groupby(columnas_pivote)['Rating']
                                                     .agg(
                                                         Moda_Rating=lambda x: x.mode().iloc[0] if not x.mode().empty else pd.NA,
                                                         Conteo_Referencia='count'
                                                     ))

#? 6.3. Verificamos cobertura sobre registros invalidos para determinar quien tiene un grupo en la tabla de referencia.
invalid_with_ref = registros_invalidos.merge(
    modas_referenciales_para_determinancia_correccion, left_on=columnas_pivote, right_index=True, how='left'
)

print(f"\nCorregibles: {invalid_with_ref['Moda_Rating'].notna().sum()} de {len(registros_invalidos)}")
print(f"Sin grupo de referencia: {invalid_with_ref['Moda_Rating'].isna().sum()}")

#? 6.4. Vista previa de los grupos de referencia
modas_referenciales_para_determinancia_correccion.sort_values('Conteo_Referencia', ascending=False)
```

```python
#? 3. Visualizacion estatica de Matplotlib para evaluar la evolucion de las ventas totales por plataforma y decada en donde algunas consolas desapareceran por decada mostrando la
#? evolucion de las ventas y consolas .


#? 3.1 Seleccionamos las top 5 plataformas en general, es decir las que mas registros tienen
top_platforms_for_trend = (
    data_cleaned['Platform']
    .value_counts()
    .head(10)
    .index
    .tolist()
)

#? 3.2 Relacionamos la tendencia de ventas de las top cinco plataformas
sales_trend = (
    data_cleaned[data_cleaned['Platform'].isin(top_platforms_for_trend)]
    .groupby(['Game_Decade', 'Platform'], observed=False)['Total_Sales']
    .sum()
    .reset_index(name='Total_Sales')
)


decades = data_cleaned['Game_Decade'].cat.categories.tolist()
n_decades = len(decades)
n_platforms = len(top_platforms_for_trend)
bar_width = 0.8 / n_platforms

fig, ax = plt.subplots(figsize=(18, 9), dpi=300)

#? 3.3 Usamos la funcion tab20 para generar una paleta de colores en este estilo
colors = plt.cm.tab20(range(n_platforms))


#? 3.4 Por cada indice y plataforma en el top 10 seleccionado, tomamos los datos y extraemos el valor de las ventas incluyendo la decada del ssales_trend
for i, platform in enumerate(top_platforms_for_trend):

    #? 3.4.1 Tomamos la informacion de Decada y Total Sales para cada paltaforma en analisis
    platform_data = sales_trend[sales_trend['Platform'] == platform]

    #? 3.4.2 Calculamos las posiciones en x para cada decada registrada con respecto del numero de decadas y la cantidad de plataaformas
    x_positions = [decades.index(d) + (i - n_platforms / 2) * bar_width + bar_width / 2
                   for d in platform_data['Game_Decade']]
    values = platform_data['Total_Sales'].tolist()

    #? 3.4.3 Definimos las barras en e eje con las posiciones calculadas y con los valores requeridos por cada paltaforma y por cada decada
    bars = ax.bar(
        x_positions,
        values,
        width=bar_width,
        label=platform,
        color=colors[i],
        edgecolor='white',
        linewidth=0.3
    )

    #? 3.4.4 Asignamos un titulo especifico a cada barra si su altura es mayor o igual que cero para identificar las plataformas que no tuvieron ventas en otros anios
    for bar in bars:
        height = bar.get_height()
        if height >= 0:
            label = f'{height / 1000:.1f}K' if height >= 1000 else f'{height:.0f}'
            ax.text(bar.get_x() + bar.get_width() / 2, height + 2,
                    label, ha='center', va='bottom', fontsize=5, fontweight='bold')

ax.set_xticks(range(n_decades))
ax.set_xticklabels(decades, fontsize=12)
ax.set_title('Evolución de Ventas Totales por Plataforma y Década (Top 10)',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Década', fontsize=12)
ax.set_ylabel('Ventas Totales (Millones USD)', fontsize=12)
ax.legend(title='Plataforma', bbox_to_anchor=(1.02, 1), loc='upper left',
          borderaxespad=0, fontsize=9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.tight_layout()
plt.show()
```

```python
#? 4. Visualizacion con Plotly de Doughnut chart responsive a cambios en el selector de plataformas
#?    para evaluar el porcentaje de ventas por region para cada plataforma a lo largo de su vida util

#? 4.1 Definimos las ventas regionales segmentadas por plataforma. La idea general es que los datos
#?     muestran la venta promedio para cada region por cada plataforma a lo largo de toda su vida util,
#?     lo que nos muestra una vision general de la distribucion de ventas por region para la plataforma.
regional_sales_by_platform = (
    data_cleaned
    #? 4.1.1 El group by nos da la estructura de ejes requerida agrupando solo por plataforma
    .groupby('Platform')
    #? 4.1.2 Este tipo de agregacion nos permite definir una salida especifica con una funcion
    #?       en base a una columna y una funcion de agregacion. Al definir x=('y', 'mean') le
    #?       decimos a pandas que realice una agregacion por promedio para las columnas de porcentaje
    .agg(
        Pct_NA=('Pct_Na_Sales', 'mean'),
        Pct_EU=('Pct_Eu_Sales', 'mean'),
        Pct_JP=('Pct_Jp_Sales', 'mean'),
        Pct_Other=('Pct_Other_Sales', 'mean')
    )
    #? 4.1.3 Finalmente hacemos un reset index para transicionar a un indice numerico
    .reset_index()
)

#? 4.2 Definimos todas las plataformas para el selector
all_platforms = sorted(data_cleaned['Platform'].unique().tolist())

#? 4.3 Definimos labels y colores especificos para tener una organizacion y lectura clara de los datos
regions = ['Pct_NA', 'Pct_EU', 'Pct_JP', 'Pct_Other']
region_labels = ['Norteamérica', 'Europa', 'Japón', 'Otras Regiones']
region_colors = ['#2171b5', '#6baed6', '#c6dbef', '#08306b']

#? 4.4 Creamos la figura de Graph Objects directamente para tener control total sobre el Pie Chart
fig_donut = go.Figure()

#? 4.5 Creamos una trace por cada plataforma. Cada trace es un doughnut chart completo
#?     que muestra la distribucion porcentual de ventas por region para esa plataforma
for idx, platform in enumerate(all_platforms):
    #? 4.5.1 Obtenemos los datos de ventas para la plataforma actual
    row = regional_sales_by_platform[
        regional_sales_by_platform['Platform'] == platform
    ]

    #? 4.5.2 Si tenemos datos y la suma de porcentajes es mayor que cero, usamos los valores reales.
    #?       Si no, asignamos ceros que no se veran en el grafico final
    if not row.empty and row[regions].sum(axis=1).values[0] > 0:
        values = [row[r].values[0] for r in regions]
    else:
        values = [0, 0, 0, 0]

    #? 4.5.3 Creamos el trace como un grafico de pastel con hueco central (doughnut).
    #?       Solo la primera plataforma es visible inicialmente
    fig_donut.add_trace(
        go.Pie(
            labels=region_labels,
            values=values,
            hole=0.5,
            marker=dict(colors=region_colors),
            textinfo='label+percent',
            textposition='outside',
            name=platform,
            visible=(idx == 0)
        )
    )

#? 4.6 Definimos los botones del selector de plataforma. Cada boton activa la visibilidad
#? de una sola trace (la correspondiente a esa plataforma) y oculta todas las demas
platform_buttons = []
for j, platform in enumerate(all_platforms):
    #? 4.6.1 Creamos un arreglo de visibilidad donde solo la trace de la plataforma seleccionada es True
    visibility = [False] * len(all_platforms)
    visibility[j] = True
    platform_buttons.append(
        dict(
            label=platform,
            method='update',
            args=[
                {'visible': visibility},
                #  args=[{}, {"title": {"text": "Initial Title"}}]
                {'title': {'text': f'Distribución Regional de Ventas — {platform}'}},
            ]
        )
    )

#? 4.7 Configuramos el layout final con el selector de plataforma como un dropdown
fig_donut.update_layout(
    title=f'Distribución Regional de Ventas',
    height=650,
    width=1200,
    updatemenus=[
        #? Aqui le decimos que menus con botones tiene que mantenner en estado el programa, especificamente le
        #? damos todos los botones de las plataformas y le decimos que muestre el boton seleccionado y que los
        #? muestre hacia abajo.
        dict(
            buttons=platform_buttons,
            direction='down',
            showactive=True,
            x=-0.1,
            xanchor='left',
            y=0.5,
            yanchor='top',
            pad=dict(r=10, t=10),
            bgcolor='#f0f0f0',
            font=dict(size=11),
            active=1
        )
    ],
    #? Aqui le decimos como tiene que mantener la referencia del texto encima del selector.
    annotations=[
        dict(
            text='Plataforma:',
            x=-0.1,
            y=0.5,
            xref='paper',
            yref='paper',
            showarrow=False,
            font=dict(size=12, color='black')
        )
    ]
)


fig_donut.show()
```

### Análisis III: Relación entre calificaciones y ventas

```python
#? 5. Relación entre calificaciones y ventas

#? 5.1 Preparamos los datos para el análisis de correlación entre calificaciones y ventas
#? Filtramos solo los juegos que tienen tanto calificaciones como ventas válidas
correlation_data = (
    data_cleaned[
        (data_cleaned['Critic_Score'].notna()) &
        (data_cleaned['User_Score'].notna()) &
        (data_cleaned['Total_Sales'] > 0) &
        (data_cleaned['Year_of_Release'].notna()) &
        (data_cleaned['Platform'].notna())
    ].copy()
)

#? 5.2 Creamos subplots lado a lado para comparar ambas correlaciones. En este caso este objeto retorna una Graph Object.Figure que podemos
#? usar para asignar trazas a diferentes partes del grafico, o a zonas de dibujo. En este aso definimos una fila pero dos columnas para explorar
#? los datos de las ventas vs calificaciones lado a lado, con lso titulos corrrespondientes y que compartan el eje y dado que la escala de datos de ventas se aplica para los mismos dos graficos
#? aunque la escala del eje x cambie.
fig_correlation = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Ventas vs Calificación de Críticos', 'Ventas vs Calificación de Usuarios'),
    horizontal_spacing=0.1, shared_yaxes=True
)

#? 5.3 Obtenemos los años únicos para el slider y los ordenamos, esto lo hacemos para tener los anios que vamos a representar en todas las trazas, dado que en este aso 
#? las trazas. Una traza es una capa de dibujo, un grafico de plotly que se coloca en un grafico, en este caso en una parte del subplot que puede ser una o varias (en este caso se precalculan)
#? varias para tener todos los datos de los dos sets, que crean un grafico. Cada traza se asigna a una parte del subplot, y luego se controla su visibilidad mediante el slider. En este caso, 
#? cada año tiene dos trazas (una para críticos y otra para usuarios) que se muestran u ocultan juntas.
years = sorted(correlation_data['Year_of_Release'].unique())
print(f"Años disponibles: {len(years)} años desde {int(years[0])} hasta {int(years[-1])}")

#? 5.4 Creamos las trazas para cada año - PRIMERO todas las de críticos, LUEGO todas las de usuarios
for i, year in enumerate(years):
    
    #? 5.4.1 Aqui obtenemos la data del set general relacionada con el anio de analisis
    year_data = correlation_data[correlation_data['Year_of_Release'] == year]

    #? 5.4.2 Creamos una traza en base a los datos de anio y critic score, que va a la derecha del grafico. Usamos la funcion
    #? go.Scatter que es parte de Graph Objects para anadir un grafico de scatter plot en la visualizacion 
    fig_correlation.add_trace(
        go.Scatter(
            x=year_data['Critic_Score'],
            y=year_data['Total_Sales'],
            mode='markers',
            name=f'Reviews de Criticos vs Ventas Totales en  <b>{int(year)}</b>',
            text=year_data['Name'],
            hovertemplate='<b>%{text}</b><br>' +
                     'Calificación Usuarios: %{x}<br>' +
                     'Ventas Totales: $%{y:.2f}M<br>' +
                     'Plataforma: %{customdata}<br>' +
                     f'Año: {int(year)}<extra></extra>',
            marker=dict(
                size=8,
                opacity=1,
                #? Aqui trabajamos para asignarle un color a cada punto representativ de su clase. El set de colores incluido en plotly que usamos corresponde a px.colors.qualitative.Set3 que 
                #? define una serie de colores que podemos usar. Para indexar y obtener un color especifico para cada plataforma, usamos el hash de la plataforma y el modulo del largo de la paleta de 
                #? colores para asegurar que cada plataforma tenga un color consistente en todas las trazas, aunque no sean contiguas.
                color=[px.colors.qualitative.Set3[hash(platform) % len(px.colors.qualitative.Set3)] for platform in year_data['Platform']],
            ),
            customdata=year_data['Platform'],

            visible=(i == 0)  # Solo el primer año visible inicialmente
        ),
        row=1, col=1
    )


for i, year in enumerate(years):
    year_data = correlation_data[correlation_data['Year_of_Release'] == year]

    fig_correlation.add_trace(
        go.Scatter(
            x=year_data['User_Score'],
            y=year_data['Total_Sales'],
            mode='markers',
            name=f'Usuarios {int(year)}',
            text=year_data['Name'],
            hovertemplate='<b>%{text}</b><br>' +
                     'Calificación Usuarios: %{x}<br>' +
                     'Ventas Totales: $%{y:.2f}M<br>' +
                     'Plataforma: %{customdata}<br>' +
                     f'Año: {int(year)}<extra></extra>',
            marker=dict(
                size=8,
                opacity=1,
                color=[px.colors.qualitative.Set3[hash(platform) % len(px.colors.qualitative.Set3)] for platform in year_data['Platform']],
            ),
            customdata=year_data['Platform'],
            visible=(i == 0)  # Solo el primer año visible inicialmente
        ),
        row=1, col=2
    )

#? 5.5 Configuramos los steps del slider - uno por cada año. En este caso creamos una por cada ano dado que por cada step del slider, Potly va a usar los datos de la vsibilidad 
#? de todo el array de anos y las trazas para filtrar y esconder aquellas con False y dejar ver aquellas con True, que por cada anio corresopnden solamente a las especificas 
#? de cada anio, En este caso, cada paso del step tiene asignado un titulo especifico y un arreglo de visibilidad que filtra cada traza de los graficos.
steps = []
for i, year in enumerate(years):
    # Crear array de visibilidad: False para todas las trazas
    visibility = [False] * (len(years) * 2)

    # Para el año i:
    # - La traza de críticos está en el índice i
    # - La traza de usuarios está en el índice i + len(years)
    visibility[i] = True  
    visibility[i + len(years)] = True  

    step = dict(
        method="update",
        args=[
            {"visible": visibility},
            {"title": f"Relación entre Calificaciones y Ventas - Año {int(year)}<br><sub>Análisis de correlación entre puntuaciones de críticos/usuarios y ventas por año</sub>"}
        ],
        label=str(int(year))
    )
    steps.append(step)

print(f"Total de trazas creadas: {len(years) * 2}")
print(f"Total de steps en slider: {len(steps)}")
print(f"Primer año: {int(years[0])}, Último año: {int(years[-1])}")

#? 5.6 Configuramos el layout con el slider
#? aqui definimos el titulo del grafico y con el estilo de datos necesarios, ademas de los sliders conectados a la visualizacion. Update Layout es una funcion 
#? que permite realizar estas modificaciones al grafico general.
fig_correlation.update_layout(
    title={
        'text': f'Relación entre Calificaciones y Ventas - Año {int(years[0])}<br><sub>Análisis de correlación entre puntuaciones de críticos/usuarios y ventas por año</sub>',
        'font': {'size': 20},
        'x': 0.5,
        'xanchor': 'center'
    },
    height=600,
    width=1500,
    showlegend=False,
    sliders=[{
        "active": 0,
        "currentvalue": {"prefix": "Año: "},
        "pad": {"t": 50},
        "steps": steps
    }]
)

#? 5.c.7 Actualizamos los ejes
fig_correlation.update_xaxes(title_text="Calificación de Críticos (0-100)", row=1, col=1)
fig_correlation.update_xaxes(title_text="Calificación de Usuarios (0-10)", row=1, col=2)
fig_correlation.update_yaxes(title_text="Ventas Totales (Millones USD)", row=1, col=1)

fig_correlation.show()
```

### Análisis IV: Comparación de ventas del mismo juego entre diferentes plataformas

```python
#? 6. Comparación de ventas del mismo juego entre plataformas

#? 6.1 Identificamos juegos que aparecen en múltiples plataformas: para hacer esto agrupamos en base al nombre de los videojuegos, que en algunos casos tienen duplicados qeu se agrupan juntos 
#? al ser de diferentes plataformas. Usamos el .nunique() de las plataformas para contar la cantidad de plataformas distintas en las que se ha registrado cada videojuego, y filtramos 
#? aquellos que tienen mas de 2 plataformas para analizarlos. Al final ordenamos en base de plataformas, que en este caso seria descendiente en el conteo de Plataformas
multi_platform_games = (
    data_cleaned
    .groupby('Name')['Platform']
    .nunique() #? 6.1.1 Conteo de plataformas unicas
    .reset_index() #? 6.1.2 Eliminamos multi-index y reducimos a columnas identificables
    .query('Platform >= 2')  #?6.1.3 Juegos en al menos 2 plataformas
    .sort_values('Platform', ascending=False)
)

#? 6.2 Obtenemos los datos de ventas para estos juegos multiplataforma: aqui filtramos la data que obtenemos en base de aquellos juegos que tienen mas de una plataforma, lo que reduce
#? nuestro dataset de 14K a 2K entradas en donde hay mas de una plataforma por juego, esto nos da toda la data de las ventas para poder usar Total Sales
multi_platform_sales = (
    data_cleaned[
        data_cleaned['Name'].isin(multi_platform_games['Name']) &
        (data_cleaned['Total_Sales'] > 0) &
        (data_cleaned['Year_of_Release'].notna())
    ].copy()
)

#? 6.3 Seleccionamos los top 25 juegos con más plataformas para mejor visualización dado que tendriamos 2400 juegos que mostrar y no se entenderia muy bien 
#? para una visualizacion de este estilo
top_multi_games = multi_platform_games.head(25)['Name'].tolist()
filtered_multi_sales = multi_platform_sales[
    multi_platform_sales['Name'].isin(top_multi_games)
].sort_values(by=['Year_of_Release','Total_Sales'],ascending=[True, False])

#? 6.4 Creamos el gráfico de barras agrupadas por juego y plataforma: Para una mejor visualizacion cambiamos la forma a barras apiladas verticales, en donde mostramos 
#? las ventas totales y segmentamos por colores en base a la plataforma 
fig_multi_platform = px.bar(
    filtered_multi_sales,
    x='Total_Sales',
    y='Name',
    color='Platform',
    animation_frame='Year_of_Release',
    title='Comparación de Ventas del Mismo Juego entre Plataformas',
    labels={
        'Total_Sales': 'Ventas Totales (M USD)',
        'Name': 'Nombre del Juego',
        'Platform': 'Plataforma',
        'Year_of_Release': 'Año'
    },
    text='Total_Sales',
    orientation='h'
)


#? 6.5 Configuramos el formato del texto y layout
fig_multi_platform.for_each_trace(
    lambda trace: trace.update(
        legendgroup=trace.name,
        showlegend=True,
        textposition='inside'
    )
)

fig_multi_platform.update_layout(
    title={
        'text': 'Comparación de Ventas del Mismo Juego entre Plataformas<br><sub>Análisis temporal de rendimiento de juegos multiplataforma</sub>',
        'font': {'size': 20},
        'x': 0.11,
        'xanchor': 'left'
    },
    xaxis_tickangle=-45,
    height=1200,
    width=2000,
    font=dict(size=10),
    margin={'autoexpand': True},
    legend=dict(
        orientation="h",
        yanchor="top",
        y=1.04,
        xanchor="left",
        x=0.4,
        title_text="",
        groupclick='toggleitem'
    )
)

fig_multi_platform.show()


```
